```python
import pandas as pd
from scipy import stats

Maledf = pd.read_csv("shot-put men.csv")
Femaledf = pd.read_csv("shot-put women.csv")

MarkMen = "Mark"

while True:
    gender = input("are you m or f?").lower()

    if gender == "m":
        while True:
            userinputX = input("Enter your mark(meters): ")
            try:
                user_value = float(userinputX)
                break
            except ValueError:
                print("Error: please enter a number.")

        rowdata = Maledf["Mark"]
        percentileX = stats.percentileofscore(rowdata, user_value, kind="weak")
        print(f"Percentile Rank for Mark: {percentileX:.2f}%")
        break

    elif gender == "f":
        while True:
            userinputY = input("Enter your mark(meters): ")
            try:
                user_value = float(userinputY)
                break
            except ValueError:
                print("Error: please enter a number.")
                
        rowdata = Femaledf["Mark"]
        percentileX = stats.percentileofscore(rowdata, user_value, kind="weak")
        print(f"Percentile Rank for Mark: {percentileX:.2f}%")
        break

    else:
        print("Invalid input. Please enter 'm' or 'f'")
```
